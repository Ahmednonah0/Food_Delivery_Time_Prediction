# Data Preprocessing for ML

- 1) Split Data into Input & Output Features

```python
x = df.drop('actual_delivery_time_minutes', axis= 1)
y = df['actual_delivery_time_minutes']
```

### Numerical Pipeline

```python
num_cols = x.select_dtypes(include= 'number').columns
num_cols
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler

median_impute = SimpleImputer(strategy= 'median')
rc = RobustScaler()

num_pipeline = Pipeline(steps= [('Median Impute', median_impute),
                                ('Scaling', rc)])
num_pipeline
```

### Categorical Pipeline

- BE Pipeline

```python
BE_cols = ['delivery_area']
BE_cols
```

```python
from category_encoders import BinaryEncoder

const_imputer = SimpleImputer(strategy= 'constant', fill_value= 'Other')
BE = BinaryEncoder()

BE_pipeline = Pipeline(steps= [('Const Impute', const_imputer),
                               ('Binary Encoder', BE)  ])
BE_pipeline
```

- Ordinal Pipeline

```python
ord_cols = ['traffic_level', 'weather']
ord_cols
```

```python
from sklearn.preprocessing import OrdinalEncoder

ord = OrdinalEncoder(categories= [
                                   ['Low', 'Medium', 'High'],
                                   ['Clear', 'Cloudy', 'Rain', 'Storm'] ])

ord_pipeline = Pipeline(steps= [ ('Ordinal Encoder', ord) ])
ord_pipeline
```

- OHE Pipeline

```python
ohe_cols = x.select_dtypes(include='object').columns.drop(['traffic_level', 'weather'])
ohe_cols
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

ohe_pipeline = Pipeline(steps= [('OHE', ohe)])
ohe_pipeline
```

### Assign each column to its corresponding Pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(transformers= [
                                                ('Num Pipeline', num_pipeline, num_cols),
                                                ('BE Pipeline', BE_pipeline, BE_cols),
                                                ('Ordinal Pipeline', ord_pipeline, ord_cols),
                                                ('OHE Pipeline', ohe_pipeline, ohe_cols)
])
preprocessing
```

### Test adding a model to the pipeline

```python
from sklearn.linear_model import LinearRegression
model_pipeline = Pipeline(steps= [('Preprocessing', preprocessing),
                                  ('Model', LinearRegression())])
model_pipeline
```

### Select Best Model

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso 
from sklearn.neighbors import KNeighborsRegressor 
from sklearn.tree import DecisionTreeRegressor 
from sklearn.ensemble import RandomForestRegressor 
from xgboost import XGBRegressor 
from sklearn.model_selection import cross_validate 
 
models = [ 
    ('Linear Regression', LinearRegression()), 
    ('Ridge', Ridge(random_state=42)), 
    ('Lasso', Lasso(random_state=42)), 
    ('KNN', KNeighborsRegressor()), 
    ('Decision Tree', DecisionTreeRegressor(random_state=42)), 
    ('Random Forest', RandomForestRegressor(
        n_estimators=200,
        max_depth=20,
        random_state=42
    )), 
    ('XGBoost', XGBRegressor(
        n_estimators=200,
        max_depth=5,
        random_state=42
    )) 
] 
 
for model in models: 
 
    model_pipeline = Pipeline(steps=[
        ('Preprocessing', preprocessing), 
        ('Model', model[1])
    ]) 
 
    cv_result = cross_validate(
        estimator=model_pipeline, 
        X=x, 
        y=y, 
        scoring=['r2', 'neg_mean_absolute_error'], 
        cv=5, 
        return_train_score=True 
    ) 
 
    print(model[0]) 
    print('Train R2 :', cv_result['train_r2'].mean().round(4)) 
    print('Test R2 :', cv_result['test_r2'].mean().round(4)) 
    print('Test MAE :', -cv_result['test_neg_mean_absolute_error'].mean().round(4)) 
    print('\n')
```

### Best Model

```python
final_pipeline = Pipeline(steps= [('Preprocessing', preprocessing),
                                      ('Model',XGBRegressor(
        n_estimators=200,
        max_depth=5,
        random_state=42
    )    ) ] )
final_pipeline
```

```python
final_pipeline.fit(x, y)
```
